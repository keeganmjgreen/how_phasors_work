```python
import matplotlib.pyplot as plt
import schemdraw
import schemdraw.elements as elm

from src.figure_utils import process_figure, rm
from src.drawing_utils import clear_axes
from src.plotting_utils import configure_matplotlib
from src.schemdraw_utils import configure_schemdraw

configure_matplotlib()
configure_schemdraw()
```

```python
fig, ax = plt.subplots(figsize=(6.4, 3), layout="tight")
clear_axes(ax)
ax.set_ylim((-4, 1.5))

with schemdraw.Drawing(canvas=ax) as d:
    elm.style(elm.STYLE_IEC)
    elm.Dot(open=True).label("$V_i$")
    I_ik = elm.Line().right()
    elm.CurrentLabelInline(ofst=0).at(I_ik).label("$I_{ik}$")  # type: ignore
    Y_S = elm.Resistor().right().label(r"$y_{ik}$")
    I_ki_n = elm.Line().right()
    elm.Dot(open=True).label("$V_{k}$")
    elm.CurrentLabelInline(ofst=0).at(I_ki_n).label(
        "$I_{ki}$"
    ).reverse()  # type: ignore
    elm.Dot().at(Y_S.start)
    elm.Resistor().down().label(r"$y_{ik}^\mathrm{Sh} \! / 2$")
    elm.Ground()
    elm.Dot().at(Y_S.end)
    elm.Resistor().down().label(
        r"$y_{ik}^\mathrm{Sh} \! / 2$", loc="bottom"
    )
    elm.Ground()
    elm.style(elm.STYLE_IEEE)

process_figure(fig, 7, 3)
```

```python
fig, ax = plt.subplots(figsize=(6.4, 3.0), layout="tight")
clear_axes(ax)
ax.set_ylim((-4, 1.5))

with schemdraw.Drawing(canvas=ax) as d:
    elm.style(elm.STYLE_IEC)
    elm.Dot(open=True).label("$V_i$", ofst=0.6).label(
        rm("Bus $i$")
    )
    I_ik = elm.Line().right()
    elm.CurrentLabelInline(ofst=0).at(I_ik).label("$I_{ik}$")  # type: ignore
    elm.Line().down(0.7)
    elm.Line().right(0.7)
    tr = (
        elm.Transformer()
        .right()
        .flip()
        .label(r"$a_{ik} \! : \! 1$")
    )
    elm.Line().at(tr.p1).left(0.7)  # type: ignore
    elm.Line().down(0.7)
    elm.Ground()
    elm.Line().at(tr.s2).right(0.7)  # type: ignore
    elm.Line().up(0.7)  # type: ignore
    I_ik_p = elm.Line().right()
    elm.CurrentLabelInline(ofst=0).at(I_ik_p).label("$I_{ik}'$")  # type: ignore
    Y_S = elm.Resistor().right().label(r"$y_{ik}$")
    I_ki_n = elm.Line().right()
    elm.Dot(open=True).label("$V_k$", ofst=0.6).label(
        rm("Bus $k$")
    )
    elm.CurrentLabelInline(ofst=0).at(I_ki_n).label(
        "$I_{ki}$"
    ).reverse()  # type: ignore
    elm.Line().at(tr.s1).right(0.7)  # type: ignore
    elm.Line().down(0.7)
    elm.Ground()
    elm.Dot().at(Y_S.start).label("$V_i'$", ofst=0.6).label(
        "$i'$"
    )
    elm.Resistor().down().label(r"$y_{ik}^\mathrm{Sh} \! / 2$")
    elm.Ground()
    elm.Dot().at(Y_S.end).label("$k'$")
    elm.Resistor().down().label(
        r"$y_{ik}^\mathrm{Sh} \! / 2$", loc="bottom"
    )
    elm.Ground()
    elm.style(elm.STYLE_IEEE)

process_figure(fig, 7, 4)
```

```python
fig, (ax1, ax2) = plt.subplots(
    ncols=2, figsize=(6.4, 2.4), layout="tight"
)
for ax in [ax1, ax2]:
    clear_axes(ax)
    ax.set_ylim((-4, 1.5))

with schemdraw.Drawing(canvas=ax1) as d:
    elm.style(elm.STYLE_IEC)
    elm.Dot(open=True).label(r"$1 \angle \delta_i$")
    elm.Line().right(2)
    elm.Line().down(0.7)
    elm.Line().right(0.7)
    tr = (
        elm.Transformer()
        .right()
        .flip()
        .label(r"$1 \! : \! 1 \angle \Phi_{ik}$")
    )
    elm.Line().at(tr.p1).left(0.7)  # type: ignore
    elm.Line().down(0.7)
    elm.Ground()
    elm.Line().at(tr.s2).right(0.7)  # type: ignore
    elm.Line().up(0.7)  # type: ignore
    elm.Line().right(1)
    Y_S = elm.Resistor().right().label(r"$-j / B_{ik}$")
    elm.Line().right(1)
    elm.Dot(open=True).label(r"$1 \angle \delta_k$")
    elm.Line().at(tr.s1).right(0.7)  # type: ignore
    elm.Line().down(0.7)
    elm.Ground()
    elm.style(elm.STYLE_IEEE)

with schemdraw.Drawing(canvas=ax2) as d:
    elm.Dot(open=True).label("$V_i$")
    elm.Line().right(2)
    elm.BatteryCell().label(r"${\Delta V}_{\! ik}$")
    elm.Line().right(1)
    Y_S = elm.Resistor().right().label(r"$r_{ik}$")
    elm.Line().right(1)
    elm.Dot(open=True).label("$V_k$")

process_figure(fig, 7, 6)
```

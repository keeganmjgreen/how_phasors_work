````python
import warnings

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.axes import Axes

warnings.filterwarnings("ignore", category=UserWarning)

plt.rcParams.update({"text.usetex": True})

xs = np.linspace(0, 2 * np.pi, 500)
````

# Power Factor and the Power Triangle

In an AC circuit with a purely resistive load, the power dissipated by the resistor---the rate at which electrical energy is converted to energy of a different form---is:

$$ p(t) = v(t) \, i(t) = \frac{v^2(t)}{R} $$

$p(t)$ is also a sinusoid that is in phase with the voltage and current. This is because $v(t)$ can be expressed as $V \! \cos x$, and $\sin^2 x$ is itself also a sinusoid through the trigonometric identity $\cos^2 x = (\cos 2 x + 1) / 2$. Unlike the voltage and current, $p(t)$ does not oscillate around $0$ and instead is always greater than or equal to zero. Furthermore, $p(t)$ oscillates with twice the frequency of the voltage and current.

````python
def format_axis(ax: Axes, x_texts: bool = True) -> None:
    if x_texts:
        ax.set_xlabel(r"$2 \pi \! f t + \theta$")
    ax.set_xlim([0, 2 * np.pi])
    ax.set_xticks([0, np.pi / 2, np.pi, 3 * np.pi / 2, 2 * np.pi])
    if x_texts:
        ax.set_xticklabels([r"$0$", r"$\pi / 2$", r"$\pi$", r"$3 \pi / 2$", r"$2 \pi$"])
    else:
        ax.set_xticklabels([])
    ax.set_yticks([0], ["$0$"])
    ax.grid(linestyle="--", axis="x")
    ax.legend(loc="upper right")


V = 5
R = 2
v = lambda x: V * np.sin(x)  # noqa: E731
i = lambda x: v(x) / R  # noqa: E731
p = lambda x: v(x) * i(x)  # noqa: E731
fig, ax = plt.subplots(1, figsize=(5.5, 4.125))
ax.plot(xs, v(xs), label="$v(t)$", color="#1565C0")
ax.plot(xs, i(xs), label="$i(t)$", color="#EF6C00")
ax.plot(xs, p(xs), label="$p(t)$", color="#2E7D32")
format_axis(ax)
ax.set_title(r"\rm Figure 4.1: Resistive load")
fig.savefig("img/fig_4_1.png", dpi=200)
````

However, in an AC circuit with inductive or capacitive components, not only do the voltage and current have different phases---the power has a different phase as well. Inductors and capacitors store energy, and the rate at which the component stores and releases electrical energy is what we call its electric power. When an AC voltage is applied to these components, they store and release energy in a cycle. Their power is a sinusoidal wave whose frequency $f$ matches that of the AC voltage. However, the times at which these components store and release energy---and the power at which they do so---does not align with the peaks and valleys of the AC voltage. The sinusoid of an inductor's and capacitor's AC power leads and lags behind that of the AC voltage by exactly an eighth of a cycle, respectively. As for a purely resistive load, the frequency of the AC power is twice that of the AC voltage. However, unlike the AC power dissipated by a resistive load, the AC power of an inductor or capacitor oscillates around $0$. The average power going into the component is equal to the average power going out of it over time. Energy is moving back and forth between the AC voltage source and the inductor or capacitor, but no energy is being dissipated.

````python
V = 5
R = 2
v = lambda x: V * np.sin(x)  # noqa: E731
i_C = lambda x: v(x - np.pi / 2) / R  # noqa: E731
p_C = lambda x: v(x) * i_C(x)  # noqa: E731
i_L = lambda x: v(x + np.pi / 2) / R  # noqa: E731
p_L = lambda x: v(x) * i_L(x)  # noqa: E731
fig, (ax1, ax2) = plt.subplots(2, figsize=(5.5, 4.125), layout="tight")
ax1.plot(xs, v(xs), label="$v(t)$", color="#1565C0")
ax1.plot(xs, i_C(xs), label="$i_C(t)$", color="#EF6C00")
ax1.plot(xs, p_C(xs), label="$p_C(t)$", color="#2E7D32")
format_axis(ax1, x_texts=False)
ax1.set_title(r"\rm Inductive load")
ax2.plot(xs, v(xs), label="$v(t)$", color="#1565C0")
ax2.plot(xs, i_L(xs), label="$i_L(t)$", color="#EF6C00")
ax2.plot(xs, p_L(xs), label="$p_L(t)$", color="#2E7D32")
format_axis(ax2)
ax2.set_title(r"\rm Capacitive load")
fig.suptitle(r"\rm Figure 4.2")
fig.savefig("img/fig_4_2.png", dpi=200)
````

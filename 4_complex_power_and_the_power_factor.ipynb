# Complex Power and the Power Factor

In the previous two chapters, we learned how complex numbers can simplify calculations involving voltage and current. We will see how complex numbers can also provide simplified power calculations. First, however, we will see how power can be derived and quantified for a generic AC circuit in the time domain in order to understand the nuances of AC power.

## Power derivations in the time domain

Consider an AC circuit with voltage source $v_S(t) = V_{\! S} \cos(\omega t + \theta_S)$ and impedance $Z = R + j X$:

````python
fig, ax = plt.subplots(figsize=(4, 3), layout="tight")
ax.set_aspect("equal")
ax.set_axis_off()
with schemdraw.Drawing(canvas=ax) as d:
    d.config(lw=1)
    v_s = elm.SourceSin().down(4).label(["$+$", "$v_S$", "$-$"]).flip()
    elm.CurrentLabelInline(direction="out").at(v_s).label(r"$i$")
    elm.Line().right(3)
    elm.style(elm.STYLE_IEC)
    X = (
        elm.Resistor()
        .up(2)
        .label(r"$jX$", loc="bottom")
        .flip()
    )
    elm.style(elm.STYLE_IEEE)
    R = elm.Resistor().up(2).label("$R$", loc="bottom")
    elm.Line().left(3)
fig.savefig("img/fig_4_1.png", dpi=200)
````

Using the approaches from the previous two chapters, we can determine the time-domain current through the circuit:

$$
i(t) = \frac{V_{\! S}}{\sqrt{R^2 + X^2}} \cos \! \left( \omega t + \theta_S - \arctan \frac{X}{R} \right)
$$

Thus, the power is:

$$
p(t) = v_S(t) \, i(t) = \frac{V_{\! S}^2}{\sqrt{R^2 + X^2}} \cos(\omega t + \theta_S) \cos \! \left( \omega t + \theta_S - \arctan \frac{X}{R} \right)
$$

Applying the trigonometric identity $\cos \theta_a \cos \theta_b = (\cos(\theta_a - \theta_b) + \cos(\theta_a + \theta_b)) / 2$ gives us the following expression for $p(t)$. We see that $p(t)$ is a sinusoid, just like $v_S(t)$ and $i(t)$. However, $p(t)$ oscillates with twice the frequency of the voltage and current, and does not necessarily oscillate around zero. The time-invariant part of $p(t)$ is the "centerline" around which the sinusoid oscillates, which is also equal to the sinusoid's time-averaged value denoted by $\bar{p}$.

$$
\begin{aligned}
p(t)
& = \frac{V_{\! S}^2}{\sqrt{R^2 + X^2}} \frac{1}{2} \! \left( \cos \! \left( \omega t + \theta_S - \left( \omega t + \theta_S - \arctan \frac{X}{R} \right) \! \right) + \cos \! \left( 2 (\omega t + \theta_S) - \arctan \frac{X}{R} \right) \! \right) \\
& = \frac{1}{2} \frac{V_{\! S}^2}{\sqrt{R^2 + X^2}} \cos \! \left( \arctan \frac{X}{R} \right) + \frac{1}{2} \frac{V_{\! S}^2}{\sqrt{R^2 + X^2}} \cos \! \left( 2 (\omega t + \theta_S) - \arctan \frac{X}{R} \right) \\
& = \underbrace{\frac{1}{2} \frac{V_{\! S}^2}{R^2 + X^2} R}_{\bar{p}, \ \text{time-invariant}} + \underbrace{\frac{1}{2} \frac{V_{\! S}^2}{\sqrt{R^2 + X^2}} \cos \! \left( 2 (\omega t + \theta_S) - \arctan \frac{X}{R} \right)}_\text{time-variant}
\end{aligned}
$$ (time_domain_p)

We can derive $p(t)$ for a purely resistive AC circuit as follows, by setting $X = 0$. We can see that this $p(t)$ does not oscillate around zero and instead is always greater than or equal to zero, as illustrated in Figure 4.1(a). This is *real* or *active* power---power that can do useful work---which we quantify it by its peak-to-peak value of $V_{\! S}^2 / R$.

$$
p(t)
= \underbrace{\, \frac{1}{2} \frac{V_{\! S}^2}{R}}_{\bar{p}} + \underbrace{\, \frac{1}{2} \frac{V_{\! S}^2}{R} \cos(2 (\omega t + \theta_S))}_\text{time-variant part}
$$

Similarly, we can derive $p(t)$ for an AC circuit with only inductors and capacitors as follows, by setting $R = 0$. The average power $\bar{p}$ is now implicitly zero; $p(t)$ oscillates around zero as illustrated in figures 4.1(a) and 4.1(b). Recall from [Chapter 1](1_introduction.ipynb) that inductive and capacitive components store energy. The rate at which the component stores and releases electrical energy is what we call its electric power---positive when converting *from* electrical energy and negative when converting back *to* electrical energy. When an AC voltage is applied to an inductive or capacitive component, it stores and releases energy in a cycle. The average power going into the component is equal to the average power going out of it over time. Power flows back and forth between the AC voltage source and the inductor or capacitor, but none is dissipated. This is *reactive* power---power that does not do useful work---which we quantify by its peak-to-peak value of $V_{\! S}^2 / X$.

$$
\begin{aligned}
p(t)
& = \underbrace{\frac{1}{2} \frac{V_{\! S}^2}{X} \cos \! \left( 2 (\omega t + \theta_S) + \frac{\pi}{2} \right)}_\text{time-variant}
\end{aligned}
$$

````python
def format_axis(ax: Axes, x_texts: bool = False) -> None:
    if x_texts:
        ax.set_xlabel(r"$\omega t + \theta$")
    ax.set_xlim([-np.pi / 2, 3 * np.pi / 2])
    ax.set_xticks([-np.pi / 2, 0, np.pi / 2, np.pi, 3 * np.pi / 2])
    if x_texts:
        ax.set_xticklabels([r"$- \pi / 2$", r"$0$", r"$\pi / 2$", r"$\pi$", r"$3 \pi / 2$"])
    else:
        ax.set_xticklabels([])
    ax.set_yticks([0], ["$0$"])
    ax.grid(linestyle="--", axis="x")
    ax.legend(loc="upper right")


xs = np.linspace(-np.pi / 2, 3 * np.pi / 2, 500)
V = 5
R = 2
v = lambda x: V * np.cos(x)  # noqa: E731
fig, (ax1, ax2, ax3) = plt.subplots(3, figsize=(5.5, 5.5), layout="tight")

i_R = lambda x: v(x) / R  # noqa: E731
p_R = lambda x: v(x) * i_R(x)  # noqa: E731
ax1.plot(xs, v(xs), label="$v(t)$", color="#1565C0")
ax1.plot(xs, i_R(xs), label="$i_R(t)$", color="#EF6C00")
ax1.plot(xs, p_R(xs), label="$p_R(t)$", color="#2E7D32")
format_axis(ax1)
ax1.set_title(r"\rm Figure 4.1(a): Resistive load")

i_L = lambda x: v(x + np.pi / 2) / R  # noqa: E731
p_L = lambda x: v(x) * i_L(x)  # noqa: E731
ax2.plot(xs, v(xs), label="$v(t)$", color="#1565C0")
ax2.plot(xs, i_L(xs), label="$i_L(t)$", color="#EF6C00")
ax2.plot(xs, p_L(xs), label="$p_L(t)$", color="#2E7D32")
format_axis(ax2)
ax2.set_title(r"\rm Figure 4.1(b): Inductive load")

i_C = lambda x: v(x - np.pi / 2) / R  # noqa: E731
p_C = lambda x: v(x) * i_C(x)  # noqa: E731
ax3.plot(xs, v(xs), label="$v(t)$", color="#1565C0")
ax3.plot(xs, i_C(xs), label="$i_C(t)$", color="#EF6C00")
ax3.plot(xs, p_C(xs), label="$p_C(t)$", color="#2E7D32")
format_axis(ax3, x_texts=True)
ax3.set_title(r"\rm Figure 4.1(c): Capacitive load")

fig.savefig("img/fig_4_2.png", dpi=200)
````

But how do we quantify the power for a circuit with nonzero $R$ *and* $X$? There are two components---power being dissipated, and power being exchanged. The power dissipated by the resistive load is:

$$
p_R(t)
= i^2(t) R
= \frac{V_{\! S}^2}{R^2 + X^2} \cos^2 \! \left( \omega t + \theta_S - \arctan \frac{X}{R} \right) R
$$

Applying the trigonometric identity $\cos^2 \! \theta = (\cos 2 \theta + 1) / 2$ gives us the following---the circuit's active power---whose peak-to-peak value is $V_{\! S}^2 R / (R^2 + X^2)$.

$$
p_R(t) = \underbrace{\frac{1}{2} \frac{V_{\! S}^2}{R^2 + X^2} R}_{\bar{p}_{\! R}, \ \text{time-invariant}} + \underbrace{\frac{1}{2} \frac{V_{\! S}^2}{R^2 + X^2} R \cos \! \left( 2 \! \left( \omega t + \theta_S - \arctan \frac{X}{R} \right) \! \right)}_\text{time-variant}
$$

The power exchanged by the reactive load is:

$$
\begin{aligned}
p_X(t)
& = p(t) - p_R(t) \\
& = \frac{1}{2} \frac{V_{\! S}^2}{R^2 + X^2} R + \frac{1}{2} \frac{V_{\! S}^2}{\sqrt{R^2 + X^2}} \cos \! \left( 2 (\omega t + \theta_S) - \arctan \frac{X}{R} \right) \\
& - \frac{1}{2} \frac{V_{\! S}^2}{R^2 + X^2} R - \frac{1}{2} \frac{V_{\! S}^2}{R^2 + X^2} R \cos \! \left( 2 \! \left( \omega t + \theta_S - \arctan \frac{X}{R} \right) \! \right) \\
& = \frac{1}{2} \frac{V_{\! S}^2}{R^2 + X^2} \left[ \sqrt{R^2 + X^2} \cos \! \left( 2 (\omega t + \theta_S) - \arctan \frac{X}{R} \right) - R \cos \! \left( 2 \! \left( \omega t + \theta_S - \arctan \frac{X}{R} \right) \! \right) \! \right]
\end{aligned}
$$

To the expression enclosed in square brackets, we now apply the trigonometric identity $a \cos(x + \theta_a) + b \cos(x + \theta_b) = c \cos(x + \theta_c)$, where $x = 2(\omega t + \theta_S)$, $c = \sqrt{a^2 + b^2 + 2 a b \cos(\theta_a - \theta_b)}$, and $\phi = \arctan((a \sin \theta_a + b \sin \theta_b) / (a \cos \theta_a + b \cos \theta_b))$.

$$
\begin{aligned}
c
& = \sqrt{\bigl( \sqrt{R^2 + X^2} \bigr)^2 + R^2 + 2 \sqrt{R^2 + X^2} (-R) \cos \! \left( 2 (\omega t + \theta_S) - \arctan \frac{X}{R} - 2 \left( \omega t + \theta_S - \arctan \frac{X}{R} \right) \! \right)} \\
& = \sqrt{2 R^2 + X^2 - 2 R \sqrt{R^2 + X^2} \cos \! \left( \arctan \frac{X}{R} \right)} \\
% & = \sqrt{2 R^2 + X^2 - 2 R \sqrt{R^2 + X^2} \, \frac{R}{\sqrt{R^2 + X^2}}} \\
& = \sqrt{2 R^2 + X^2 - 2 R^2} \\
& = X \\
\theta_c
& = \arctan \frac{\sqrt{R^2 + X^2} \sin \! \left( - \arctan \frac{X}{R} \right) - R \sin \! \left( -2 \arctan \frac{X}{R} \right)}{\sqrt{R^2 + X^2} \cos \! \left( - \arctan \frac{X}{R} \right) - R \cos \! \left( -2 \arctan \frac{X}{R} \right)} \\
% Apply reflection and double-angle identities:
& = \arctan \frac{- \sqrt{R^2 + X^2} \sin \! \left( \arctan \frac{X}{R} \right) + 2 R \sin \! \left( \arctan \frac{X}{R} \right) \cos \! \left( \arctan \frac{X}{R} \right)}{\sqrt{R^2 + X^2} \cos \! \left( \arctan \frac{X}{R} \right) - R \left(2 \cos^2 \! \left( \arctan \frac{X}{R} \right) - 1 \right)} \\
% & = \arctan \frac{- \sqrt{R^2 + X^2} \frac{X}{\sqrt{R^2 + X^2}} + 2 R \frac{X}{\sqrt{R^2 + X^2}} \frac{R}{\sqrt{R^2 + X^2}}}{\sqrt{R^2 + X^2} \frac{R}{\sqrt{R^2 + X^2}} - R \left(2 \left( \frac{R}{\sqrt{R^2 + X^2}} \right)^2 - 1 \right)} \\
% Simplify:
& = \arctan \frac{- X + 2 R \frac{X R}{R^2 + X^2}}{R - R \left(2 \frac{R^2}{R^2 + X^2} - 1 \right)} \\
% & = \arctan \frac{- X + 2 R \frac{X R}{R^2 + X^2}}{2 R - 2 R \frac{R^2}{R^2 + X^2}} \\
% Multiply both numerator and denominator of arctan's argument by $(R^2 + X^2)$:
& = \arctan \frac{- X (R^2 + X^2) + 2 R (X R)}{2 R (R^2 + X^2) - 2 R R^2} \\
% & = \arctan \frac{- R^2 X - X^3 + 2 R^2 X}{2 R^3 + 2 R X^2 - 2 R^3} \\
% & = \arctan \frac{R^2 X - X^3}{2 R X^2} \\
& = \arctan \frac{R^2 - X^2}{2 R X}
\end{aligned}
$$

Thus, reactive load's power is the following---the circuit's reactive power---whose peak-to-peak value is $V_{\! S}^2 X / (R^2 + X^2)$.

$$
p_X(t) = \frac{1}{2} \frac{V_{\! S}^2}{R^2 + X^2} X \cos \! \left( 2 (\omega t + \theta_S) + \arctan \frac{R^2 - X^2}{2 R X} \right)
$$

## Power as a phasor?

If $p(t) = v^2(t) / R$ for a purely resistive circuit, then it follows that the analog for a circuit including reactive components would be $\mathbf{p} = \mathbf{v}_{\! S}^2 / Z$:

$$
\begin{aligned}
\mathbf{p}
& = \frac{\mathbf{v}_{\! S}^2}{Z} \\
& = \frac{(V_{\! S} \angle \theta_S)^2}{\sqrt{R^2 + X^2} \angle \arctan \frac{X}{R}} \\
& = \frac{V_{\! S}^2 \angle 2 \theta_S}{\sqrt{R^2 + X^2} \angle \arctan \frac{X}{R}} \\
& = \frac{V_{\! S}^2}{\sqrt{R^2 + X^2}} \: \angle \left( 2 \theta_S - \arctan \frac{X}{R} \right)
\end{aligned}
$$ (naive_complex_p)

If we were to naively convert this into the time domain, we would find that it does not match Equation {eq}`time_domain_p` which we derived in the previous section for $p(t)$. Firstly, because it oscillates around zero rather than around $V_{\! S}^2 R \,\! / \, 2 (R^2 + X^2)$. Secondly, because its amplitude is $V_{\! S}^2 / \sqrt{R^2 + X^2}$ rather than $V_{\! S}^2 / \, 2 \sqrt{R^2 + X^2}$. And thirdly, because it oscillates at natural frequency $\omega$ rather than $2 \omega$. Clearly, this doesn't work---power is not a phasor.

$$
\frac{V_{\! S}^2}{\sqrt{R^2 + X^2}} \cos \! \left( \omega t + 2 \theta_S - \arctan \frac{X}{R} \right) \neq \underbrace{\frac{1}{2} \frac{V_{\! S}^2}{R^2 + X^2} R + \frac{1}{2} \frac{V_{\! S}^2}{\sqrt{R^2 + X^2}} \cos \! \left( 2 (\omega t + \theta_S) - \arctan \frac{X}{R} \right)}_\text{From Equation 4.1}
$$

## Complex power

We want to quantify the active power and reactive power of a circuit. These do not depend on $\theta_S$. If we were to treat $\theta_S$ as if it were zero in Equation {eq}`naive_complex_p`, we could write:

$$
\begin{aligned}
\mathbf{p}
& = \frac{V_{\! S}^2}{\sqrt{R^2 + X^2}} \: \angle \arctan \frac{X}{R} \\
& = \frac{V_{\! S}^2}{\sqrt{R^2 + X^2}} \cos \! \left( \arctan \frac{X}{R} \right) + j \, \frac{V_{\! S}^2}{\sqrt{R^2 + X^2}} \sin \! \left( \arctan \frac{X}{R} \right) \\
& = \frac{V_{\! S}^2}{\sqrt{R^2 + X^2}} \frac{R}{\sqrt{R^2 + X^2}} + j \, \frac{V_{\! S}^2}{\sqrt{R^2 + X^2}} \frac{X}{\sqrt{R^2 + X^2}} \\
& = \frac{V_{\! S}^2}{R^2 + X^2} R + j \, \frac{V_{\! S}^2}{R^2 + X^2} X
\end{aligned}
$$

Note how the real part $V_{\! S}^2 R / (R^2 + X^2)$ and the imaginary part $V_{\! S}^2 X / (R^2 + X^2)$ are the (peak-to-peak) values of the circuit's active and reactive power.

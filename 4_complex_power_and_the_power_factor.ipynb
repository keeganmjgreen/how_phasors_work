# Complex Power and the Power Factor

In the previous two chapters, we learned how complex numbers can simplify calculations involving voltage and current. We will see how complex numbers can also provide simplified power calculations. But first, in order to understand the nuances of AC power, we will see how power can be derived and quantified for a general AC circuit in the time domain.

## Power derivations in the time domain

Consider an AC circuit with voltage source $v_S(t) = |V_{\! S}| \cos(\omega t + \theta_S)$ and impedance $Z = R + {j} X$. Such a circuit is useful because it can be used to model any series connection between one or more impedances and one or more voltage sources---as long as the voltage sources can be lumped together as one $v_S(t)$ and the impedances can be lumped together as one $Z$.

````python
fig, ax = plt.subplots(figsize=(6, 3), layout="tight")
ax.set_aspect("equal")
ax.set_axis_off()
with schemdraw.Drawing(canvas=ax) as d:
    d.config(lw=1)
    v_s = (
        elm.SourceSin()
        .down(4)
        .label(["$+$", r"$p(t) \longrightarrow \quad v_S(t)$", "$-$"])
        .flip()
    )
    elm.CurrentLabelInline(direction="out").at(v_s).label(r"$i(t)$", loc="bottom")
    elm.Line().right(3)
    elm.style(elm.STYLE_IEC)
    X = (
        elm.Resistor()
        .up(2)
        .label(r"${j} X \quad \longrightarrow p_X(t)$", loc="bottom")
        .flip()
    )
    elm.style(elm.STYLE_IEEE)
    R = elm.Resistor().up(2).label(r"$R \quad\;\;\, \longrightarrow p_R(t)$", loc="bottom")
    elm.Line().left(3)
fig.suptitle(r"\rm \Large Figure 4.1")
fig.savefig("img/fig_4_1.png", dpi=200)
````

Using the approaches from the previous two chapters, we can determine the current through the circuit in the time domain:

$$
i(t) = \frac{|V_{\! S}|}{\sqrt{R^2 + X^2}} \cos \! \left( \omega t + \theta_S - \arctan \frac{X}{R} \right)
$$

Thus, the instantaneous power at time $t$ is:

$$
p(t) = v_S(t) \, i(t) = \frac{|V_{\! S}|^2}{\sqrt{R^2 + X^2}} \cos(\omega t + \theta_S) \cos \! \left( \omega t + \theta_S - \arctan \frac{X}{R} \right)
$$

Applying the trigonometric identity $\cos \theta_a \cos \theta_b = (\cos(\theta_a - \theta_b) + \cos(\theta_a + \theta_b)) / 2$ gives us the following expression for $p(t)$. We see that $p(t)$ is a sinusoid, just like $v_S(t)$ and $i(t)$. However, $p(t)$ oscillates with twice the frequency of the voltage and current ($2 \omega$, instead of $\omega$), and does not necessarily oscillate around zero. The time-invariant part of $p(t)$ is the "centerline" around which the sinusoid oscillates, which is also equal to the sinusoid's time-averaged value denoted by $\bar{p}$.

$$
\begin{aligned}
p(t)
& = \frac{|V_{\! S}|^2}{\sqrt{R^2 + X^2}} \\
& \times \frac{1}{2} \! \left( \cos \! \left( \omega t + \theta_S - \left( \omega t + \theta_S - \arctan \frac{X}{R} \right) \! \right) + \cos \! \left( 2 (\omega t + \theta_S) - \arctan \frac{X}{R} \right) \! \right) \\
& = \frac{1}{2} \frac{|V_{\! S}|^2}{\sqrt{R^2 + X^2}} \cos \! \left( \arctan \frac{X}{R} \right) + \frac{1}{2} \frac{|V_{\! S}|^2}{\sqrt{R^2 + X^2}} \cos \! \left( 2 (\omega t + \theta_S) - \arctan \frac{X}{R} \right) \\
& = \underbrace{\frac{1}{2} \frac{|V_{\! S}|^2}{R^2 + X^2} R}_{\bar{p}, \ \text{time-invariant}} + \underbrace{\frac{1}{2} \frac{|V_{\! S}|^2}{\sqrt{R^2 + X^2}} \cos \! \left( 2 (\omega t + \theta_S) - \arctan \frac{X}{R} \right)}_\text{time-variant}
\end{aligned}
$$ (time_domain_p)

We can derive $p(t)$ for a purely resistive AC circuit as follows, by setting $X = 0$. We can see that $p(t)$ is always greater than (or equal to) zero, as illustrated in Figure 4.2(a). Because resistor $R$ always dissipates power and in the real world likely represents a device that can do useful work---such as a light bulb or laptop---we call this *real* or *active* power. In the $X = 0$ case, we can quantify our circuit's active power by its peak-to-peak value of $|V_{\! S}|^2 / R$.

$$
p(t)
= \underbrace{\, \frac{1}{2} \frac{|V_{\! S}|^2}{R}}_{\bar{p}} + \underbrace{\, \frac{1}{2} \frac{|V_{\! S}|^2}{R} \cos(2 (\omega t + \theta_S))}_\text{time-variant part}
$$

Similarly, we can derive $p(t)$ for an AC circuit with only inductors and capacitors as follows, by setting $R = 0$. The average power $\bar{p}$ is now zero; $p(t)$ oscillates around zero as illustrated in figures 4.2(a) and 4.2(b). Recall from [Chapter 1](1_introduction.ipynb) that inductive and capacitive components store energy. The rate at which the component stores or releases electrical energy is what we call its electric power---positive when converting *from* electrical energy and negative when converting back *to* electrical energy. When an AC voltage is applied to an inductive or capacitive component, it stores and releases energy in a cycle. The average power going into the component is equal to the average power going out of it over time. Power is exchanged back and forth between the AC voltage source and the inductor or capacitor, but none is dissipated or does any useful work. We call this *reactive* power. In the $R = 0$ case, we can quantify our circuit's reactive power by its peak-to-peak value of $|V_{\! S}|^2 / X$.

$$
\begin{aligned}
p(t)
& = \underbrace{\frac{1}{2} \frac{|V_{\! S}|^2}{X} \cos \! \left( 2 (\omega t + \theta_S) + \frac{\pi}{2} \right)}_\text{time-variant}
\end{aligned}
$$

````python
def format_axis(ax: Axes, x_texts: bool = False) -> None:
    if x_texts:
        ax.set_xlabel(r"$\omega t + \theta$")
    ax.set_xlim((-np.pi / 2, 3 * np.pi / 2))
    ax.set_xticks([-np.pi / 2, 0, np.pi / 2, np.pi, 3 * np.pi / 2])
    if x_texts:
        ax.set_xticklabels([r"$- \pi / 2$", r"$0$", r"$\pi / 2$", r"$\pi$", r"$3 \pi / 2$"])
    else:
        ax.set_xticklabels([])
    ax.set_yticks([0], ["$0$"])
    ax.grid(linestyle="--", axis="x")
    ax.legend(loc="upper right")


xs = np.linspace(-np.pi / 2, 3 * np.pi / 2, 500)
V = 5
R = 2
v = lambda x: V * np.cos(x)  # noqa: E731
fig, (ax, ax2, ax3) = plt.subplots(3, figsize=(5.5, 5.5), layout="tight")

i_R = lambda x: v(x) / R  # noqa: E731
p_R = lambda x: v(x) * i_R(x)  # noqa: E731
ax.plot(xs, v(xs), label="$v(t)$", color="#1565C0")
ax.plot(xs, i_R(xs), label="$i_R(t)$", color="#EF6C00")
ax.plot(xs, p_R(xs), label="$p_R(t)$", color="#2E7D32")
format_axis(ax)
ax.set_title(r"\rm Figure 4.2(a): Resistive load")

i_L = lambda x: v(x + np.pi / 2) / R  # noqa: E731
p_L = lambda x: v(x) * i_L(x)  # noqa: E731
ax2.plot(xs, v(xs), label="$v(t)$", color="#1565C0")
ax2.plot(xs, i_L(xs), label="$i_L(t)$", color="#EF6C00")
ax2.plot(xs, p_L(xs), label="$p_L(t)$", color="#2E7D32")
format_axis(ax2)
ax2.set_title(r"\rm Figure 4.2(b): Inductive load")

i_C = lambda x: v(x - np.pi / 2) / R  # noqa: E731
p_C = lambda x: v(x) * i_C(x)  # noqa: E731
ax3.plot(xs, v(xs), label="$v(t)$", color="#1565C0")
ax3.plot(xs, i_C(xs), label="$i_C(t)$", color="#EF6C00")
ax3.plot(xs, p_C(xs), label="$p_C(t)$", color="#2E7D32")
format_axis(ax3, x_texts=True)
ax3.set_title(r"\rm Figure 4.2(c): Capacitive load")

fig.savefig("img/fig_4_2.png", dpi=200)
````

But how do we quantify the power for a circuit with nonzero $R$ *and* $X$? There are two components---power being dissipated, and power being exchanged. Firstly, the power dissipated by the resistive load is:

$$
p_R(t)
= i^2(t) R
= \frac{V_{\! S}^2}{R^2 + X^2} \cos^2 \! \left( \omega t + \theta_S - \arctan \frac{X}{R} \right) R
$$

Applying the trigonometric identity $\cos^2 \! \theta = (\cos 2 \theta + 1) / 2$ gives us the following---the circuit's active power---whose peak-to-peak value is $V_{\! S}^2 R / (R^2 + X^2)$. Note that even though this is only active power and is only dissipated by the resistor, it is affected by $X$ in addition to $R$.

$$
p_R(t) = \underbrace{\frac{1}{2} \frac{V_{\! S}^2}{R^2 + X^2} R}_{\bar{p}_{\! R}, \ \text{time-invariant}} + \underbrace{\frac{1}{2} \frac{V_{\! S}^2}{R^2 + X^2} R \cos \! \left( 2 \! \left( \omega t + \theta_S - \arctan \frac{X}{R} \right) \! \right)}_\text{time-variant}
$$

And secondly, the power exchanged by the reactive load is:

$$
\begin{aligned}
p_X(t)
& = p(t) - p_R(t) \\
& = \frac{1}{2} \frac{V_{\! S}^2}{R^2 + X^2} R + \frac{1}{2} \frac{V_{\! S}^2}{\sqrt{R^2 + X^2}} \cos \! \left( 2 (\omega t + \theta_S) - \arctan \frac{X}{R} \right) \\
& - \frac{1}{2} \frac{V_{\! S}^2}{R^2 + X^2} R - \frac{1}{2} \frac{V_{\! S}^2}{R^2 + X^2} R \cos \! \left( 2 \! \left( \omega t + \theta_S - \arctan \frac{X}{R} \right) \! \right) \\
& = \frac{1}{2} \frac{V_{\! S}^2}{R^2 + X^2} \\
& \times \left[ \sqrt{R^2 + X^2} \cos \! \left( 2 (\omega t + \theta_S) - \arctan \frac{X}{R} \right) - R \cos \! \left( 2 \! \left( \omega t + \theta_S - \arctan \frac{X}{R} \right) \! \right) \! \right]
\end{aligned}
$$

To the expression enclosed in square brackets, we now apply the trigonometric identity $a \cos(x + \theta_a) + b \cos(x + \theta_b) = c \cos(x + \theta_c)$, where $c = \sqrt{a^2 + b^2 + 2 a b \cos(\theta_a - \theta_b)}$ and $\theta_c = \arctan((a \sin \theta_a + b \sin \theta_b) / (a \cos \theta_a + b \cos \theta_b))$.

$$
\begin{aligned}
x
& = 2(\omega t + \theta_S) \\\\
c
& = \sqrt{
\begin{aligned}
& \left( \! \sqrt{R^2 + X^2} \!\: \right)^{\! 2} + R^2 \\
& + 2 \sqrt{R^2 + X^2} \, (-R) \cos \! \left( 2 (\omega t + \theta_S) - \arctan \frac{X}{R} - 2 \left( \omega t + \theta_S - \arctan \frac{X}{R} \right) \! \right)
\end{aligned}
} \\
& = \sqrt{\, 2 R^2 + X^2 - 2 R \sqrt{R^2 + X^2} \cos \! \left( \arctan \frac{X}{R} \right)} \\
% & = \sqrt{2 R^2 + X^2 - 2 R \sqrt{R^2 + X^2} \, \frac{R}{\sqrt{R^2 + X^2}}} \\
& = \sqrt{2 R^2 + X^2 - 2 R^2} \\
& = X \\
\theta_c
& = \arctan \frac{\sqrt{R^2 + X^2} \sin \! \left( - \arctan \frac{X}{R} \right) - R \sin \! \left( -2 \arctan \frac{X}{R} \right)}{\sqrt{R^2 + X^2} \cos \! \left( - \arctan \frac{X}{R} \right) - R \cos \! \left( -2 \arctan \frac{X}{R} \right)} \\
% Apply reflection and double-angle identities:
& = \arctan \frac{- \sqrt{R^2 + X^2} \sin \! \left( \arctan \frac{X}{R} \right) + 2 R \sin \! \left( \arctan \frac{X}{R} \right) \cos \! \left( \arctan \frac{X}{R} \right)}{\sqrt{R^2 + X^2} \cos \! \left( \arctan \frac{X}{R} \right) - R \left(2 \cos^2 \! \left( \arctan \frac{X}{R} \right) - 1 \right)} \\
% & = \arctan \frac{- \sqrt{R^2 + X^2} \frac{X}{\sqrt{R^2 + X^2}} + 2 R \frac{X}{\sqrt{R^2 + X^2}} \frac{R}{\sqrt{R^2 + X^2}}}{\sqrt{R^2 + X^2} \frac{R}{\sqrt{R^2 + X^2}} - R \left(2 \left( \frac{R}{\sqrt{R^2 + X^2}} \right)^2 - 1 \right)} \\
% Simplify:
& = \arctan \frac{- X + 2 R \frac{X R}{R^2 + X^2}}{R - R \left(2 \frac{R^2}{R^2 + X^2} - 1 \right)} \\
% & = \arctan \frac{- X + 2 R \frac{X R}{R^2 + X^2}}{2 R - 2 R \frac{R^2}{R^2 + X^2}} \\
% Multiply both numerator and denominator of arctan's argument by $(R^2 + X^2)$:
& = \arctan \frac{- X (R^2 + X^2) + 2 R (X R)}{2 R (R^2 + X^2) - 2 R R^2} \\
% & = \arctan \frac{- R^2 X - X^3 + 2 R^2 X}{2 R^3 + 2 R X^2 - 2 R^3} \\
% & = \arctan \frac{R^2 X - X^3}{2 R X^2} \\
& = \arctan \frac{R^2 - X^2}{2 R X}
\end{aligned}
$$

Thus, reactive load's power is the following---the circuit's reactive power---whose peak-to-peak value is $V_{\! S}^2 X / (R^2 + X^2)$. Note that even though this is only reactive power and is only exchanged by the reactive load, it is affected by $R$ in addition to $X$.

$$
p_X(t) = \frac{1}{2} \frac{V_{\! S}^2}{R^2 + X^2} X \cos \! \left( 2 (\omega t + \theta_S) + \arctan \frac{R^2 - X^2}{2 R X} \right)
$$

Now we can determine the active and reactive power for any circuit like that in Figure 4.1.

## Power as a phasor?

The previous section informed us of the nuances of AC power, but the calculations were tedious. We ought to be able to determine the active and reactive power of a circuit---or of any of its components---more easily. Perhaps if AC voltage and current calculations are simplified by working in the phasor domain, then so would AC power calculations. If $p(t) = v^2(t) / R$ for a purely resistive circuit, then surely the equivalent for a circuit that includes reactive components would be $P = V_{\! S}^2 / Z$? Let's attempt to use this definition:

$$
\begin{aligned}
P
& = \frac{V_{\! S}^2}{Z} \\
& = \frac{(|V_{\! S}| \angle \theta_S)^2}{\sqrt{R^2 + X^2} \angle \arctan \frac{X}{R}} \\
& = \frac{|V_{\! S}|^2 \angle 2 \theta_S}{\sqrt{R^2 + X^2} \angle \arctan \frac{X}{R}} \\
& = \frac{|V_{\! S}|^2}{\sqrt{R^2 + X^2}} \: \angle \left( 2 \theta_S - \arctan \frac{X}{R} \right)
\end{aligned}
$$ (naive_complex_P)

If we were to naively convert this phasor into the time domain as follows, we would find that it does not match Equation {eq}`time_domain_p` for $p(t)$ which we derived in the previous section. Firstly, because it oscillates around zero rather than around $|V_{\! S}|^2 R \,\! / \, 2 (R^2 + X^2)$. Secondly, because its amplitude is $|V_{\! S}|^2 / \sqrt{R^2 + X^2}$ rather than $|V_{\! S}|^2 / \, 2 \sqrt{R^2 + X^2}$. And thirdly, because it oscillates at natural frequency $\omega$ rather than $2 \omega$. Clearly, this does not work---power cannot be represented by a phasor.

$$
\begin{aligned}
& \frac{|V_{\! S}|^2}{\sqrt{R^2 + X^2}} \cos \! \left( \omega t + 2 \theta_S - \arctan \frac{X}{R} \right) \\
& \neq \underbrace{\frac{1}{2} \frac{|V_{\! S}|^2}{R^2 + X^2} R + \frac{1}{2} \frac{|V_{\! S}|^2}{\sqrt{R^2 + X^2}} \cos \! \left( 2 (\omega t + \theta_S) - \arctan \frac{X}{R} \right)}_\text{From Equation 4.1}
\end{aligned}
$$

However, this does not necessarily mean that AC power cannot be quantified by a complex number or that complex numbers cannot somehow be used to simplify AC power calculations. Note that active and reactive power do not depend on $\theta_S$. In Equation {eq}`naive_complex_P`, if we were to treat $\theta_S$ as if it were zero, we could write:

$$
\begin{aligned}
P
& = \frac{|V_{\! S}|^2}{\sqrt{R^2 + X^2}} \: \angle \arctan \frac{X}{R} \\
& = \frac{|V_{\! S}|^2}{\sqrt{R^2 + X^2}} \cos \! \left( \arctan \frac{X}{R} \right) + {j} \, \frac{|V_{\! S}|^2}{\sqrt{R^2 + X^2}} \sin \! \left( \arctan \frac{X}{R} \right) \\
& = \frac{|V_{\! S}|^2}{\sqrt{R^2 + X^2}} \frac{R}{\sqrt{R^2 + X^2}} + {j} \, \frac{|V_{\! S}|^2}{\sqrt{R^2 + X^2}} \frac{X}{\sqrt{R^2 + X^2}} \\
& = \frac{|V_{\! S}|^2}{R^2 + X^2} R + {j} \, \frac{|V_{\! S}|^2}{R^2 + X^2} X
\end{aligned}
$$

Note how the real part $|V_{\! S}|^2 R / (R^2 + X^2)$ and the imaginary part $|V_{\! S}|^2 X / (R^2 + X^2)$ are the (peak-to-peak) values of the circuit's active and reactive power that we derived in the previous section. This means that we are en route to a way to use complex numbers to quantify AC power. In practice, it is not the entire expression $|V_{\! S}|^2 R / (R^2 + X^2) + {j} \, |V_{\! S}|^2 X / (R^2 + X^2)$ that we denote $P$, but just the real part $|V_{\! S}|^2 R / (R^2 + X^2)$. We denote the imaginary part $|V_{\! S}|^2 X / (R^2 + X^2)$ by $Q$, and the expression as a whole by complex number $S$. Thus, $S = P + {j} \, Q$. $S$ is what we call *complex power*.

## Complex power

By quantifying AC power as

$$
\boxed{S = |V_{\! S}|^2 / Z^*}
$$ (S_eq_V_mag_sq_over_Z_conj)

rather than as $V_{\! S}^2 / Z$, we can remove $\theta_S$ from our calculations and make them simpler. For example, for a circuit like that in Figure 4.1, we can calculate complex power $S$ easily using only complex number division:

$$
S
= \frac{|V_{\! S}|^2}{R - {j} X}
= \underbrace{\frac{|V_{\! S}|^2}{R^2 + X^2} R}_P + {j} \, \underbrace{\frac{|V_{\! S}|^2}{R^2 + X^2} X}_Q
$$

From Equation {eq}`S_eq_V_mag_sq_over_Z_conj`, we can derive the following---the AC power equivalent of the formula $P = I^2 R$.

$$
S
= \frac{|V_{\! S}|^2}{Z^*}
= \frac{|V_{\! S}|^2}{Z^*} \frac{Z}{Z}
= \frac{|V_{\! S}|^2 \, Z}{|Z|^2}
= \left( \frac{|V_{\! S}|}{|Z|} \right)^{\!\! 2} Z
\implies \boxed{S = |I|^2 Z}
$$ (S_eq_I_mag_sq_times_Z)

From which we can derive the AC power equivalent of $P = V_{\! S} I$:

$$
S
= |I|^2 Z
= |I|^2 \frac{V_{\! S}}{I}
= |I|^2 \frac{V_{\! S}}{I} \frac{I^*}{I^*}
= |I|^2 \frac{V_{\! S} I^*}{|I|^2}
\implies \boxed{S = V_{\! S} I^*}
$$ (S_eq_V_times_I_conj)

## The power triangle

In [](2_what_is_a_phasor.ipynb), we saw how a triangle could be used to depict a phasor (and thus any complex number) in the complex plane. Accordingly, complex power is often depicted as a *power triangle*:

````python
fig, ax = plt.subplots(figsize=(3.5, 2.5), layout="tight")
ax.set_title(r"\rm \small Figure 4.3")

ax.axis("off")
ax.set_aspect("equal")

ax.set_xlim((-1, 4))
ax.set_ylim((-0.5, 3))

real_axis = (
    Arrow(Point(-0.5, 0), Point(3.5, 0))
    .drawn(ax)
    .end.labeled(ax, r"\rm Re", (7.5, -3))
)
imag_axis = (
    Arrow(Point(0, -0.5), Point(0, 2.5)).drawn(ax).end.labeled(ax, r"\rm Im", (0, 2.5))
)

origin = Point(0, 0).drawn(ax)
P = Point(3, 0).drawn(ax)
S = Point(3, 2).drawn(ax)
Arrow(origin, P).drawn(ax).mid.labeled(ax, r"$P$", (0, -12))
Arrow(P, S).drawn(ax).mid.labeled(ax, r"${j} \, Q$", (12, -5))
Arrow(origin, S).drawn(ax).mid.labeled(ax, r"$S$", (-5, 5))

Arc.from_vertex_start_end(vertex=origin, start=P, end=S).drawn(ax).labeled(
    ax, r"$\phi$", (0, -2.5)
)

fig.savefig("img/fig_4_3.png", dpi=200)
````

Even though power $p(t)$ is a sinusoid, and a phasor rotating in the complex plane traces out a sinusoid, $S$ is not a phasor and would not form $p(t)$ if rotated about the origin. However, if $S - P / \, 2$ were rotated about $P / \, 2$, it would correctly trace out $p(t)$.
